```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv('/Users/atulpimpalkar/Downloads/Mindlfyt AI/Salary_Data_multi.csv')
print(df.head())
```

```python
X=df[['YearsExperience', 'Age']]
y=df['Salary']
model=LinearRegression()
model.fit(X, y)
print(model.coef_)
print(model.intercept_)
```

```python
y=model.predict(X)
print(y)
```

```python
y1=model.predict([[5,30]])
print(y1)
```

```python
rmsse = ((y1 - y) ** 2).mean() ** 0.5
print(rmsse)
```

```python
y_actual=df['Salary']
y_predicted=y
rmse = ((y_actual - y_predicted) ** 2).mean() ** 0.5
print(rmse)   
```

```python
plt.scatter(df['YearsExperience'], df['Salary'], color='blue', label='Actual Salary')
plt.scatter(df['YearsExperience'], y, color='red', label='Predicted Salary')
plt.title('Actual vs Predicted Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()  
```
